```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style
style.use('fast')
experiment_ID = "Col_48h_15t_RNAseq"
number_jobs = 100
header_list = ["Cause Gene", "Effect Gene", "mean_importance"]
```

# SWING

```python
pathway_data = f"C:/Users/ivana/OneDrive/Desktop/Internship New Folder/Analysis Experimental Data/Results_instances-1000/Results_SWING_{experiment_ID}_genes-157"
                
mean_important_array_SWING = np.array([])
mean_important_array_rand_SWING = np.array([])
mean_important_array_crand_SWING = np.array([])

end = 0
for j in range(number_jobs):
    
    first = end + 1
    end = end + 10
                                                                                        
    mean_importance_file = pd.read_csv(f"{pathway_data}/Mean_Importance_{experiment_ID}/ranking_SWING_{experiment_ID}_range{first}-{end}.txt", sep = "\t")                                                                                             
    mean_importance_file_rand = pd.read_csv(f"{pathway_data}/Mean_Importance_{experiment_ID}_rand/ranking_SWING_rand_{experiment_ID}_range{first}-{end}.txt", sep = "\t")
    mean_importance_file_crand = pd.read_csv(f"{pathway_data}/Mean_Importance_{experiment_ID}_crand/ranking_SWING_crand_{experiment_ID}_range{first}-{end}.txt", sep = "\t")
    
    mean_important_array_SWING = np.append(mean_important_array_SWING, mean_importance_file[f"mean_importance"].values)
    mean_important_array_rand_SWING= np.append(mean_important_array_rand_SWING, mean_importance_file_rand[f"mean_importance"].values)
    mean_important_array_crand_SWING= np.append(mean_important_array_crand_SWING, mean_importance_file_crand[f"mean_importance"].values)
```

```python
figure = plt.figure(figsize=(8,5))
plt.rc('xtick', labelsize=17)
plt.rc('ytick', labelsize=17)
plt.rcParams["font.family"] = "serif"

y, x, _ = plt.hist(mean_important_array_SWING, bins = 200, color = "midnightblue", density =1, label = r"Causal data")
y_rand, x_rand, _ = plt.hist(mean_important_array_rand_SWING, bins = 200, color='mediumspringgreen', alpha = 0.41, density = 1, label = r"Data shuffled in time")
y_crand, x_crand, _ = plt.hist(mean_important_array_crand_SWING, bins = 200, color='k', alpha = 0.3, density = 1, label = r"Data fully shuffled")

plt.ylabel("Normalized frequency", fontsize = 19)
plt.xlabel("Rank of importance interactions", fontsize = 19)
plt.title("SWING WT 15t 157 genes", fontsize = 22)
plt.xlim(0,0.01)
plt.legend(prop={"size":15.5, "family":"serif"})
plt.text(-0.00099,910, "(A)", fontsize = 30)
plt.savefig(f"C:/Users/ivana/OneDrive/Desktop/Imagenes Internship/histogram_SWING_experimental_WT.png", bbox_inches="tight", dpi=100)

#Statistics
print("mean: ", mean_important_array_SWING.mean())
print("mean rand: ", mean_important_array_rand_SWING.mean())
print("estandard desv: ", mean_important_array_SWING.std())
print("estandard desv rand: ", mean_important_array_rand_SWING.std())

print("most frequent value: ", x.max(), " ", y.max())
print("most frequent value rand: ",  x_rand.max(), " ", y_rand.max())
```
